```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Leitura do arquivo CSV
dataframe = pd.read_csv(r'..\dados\vendas_desafio.csv', sep=',')
```

```python
# Criar coluna faturamento = quantidade * preco_unitario
dataframe['faturamento'] = round(dataframe['quantidade'] * dataframe['preco_unitario'], 2)
```

```python
# Exibir os primeiros registros para validação inicial
display(dataframe.head())
```

```python
# Agrupar faturamento por produto
dataframe.groupby('produto')['faturamento'].sum()
```

```python
# Agrupar faturamento por categoria
dataframe.groupby('categoria')['faturamento'].sum()
```

```python
# Agrupar faturamento por cliente
dataframe.groupby('cliente')['faturamento'].sum()
```

```python
# Foi criada a variável df_cat_fat para agrupar os dados antes de gerar o gráfico.
df_categ_fat = dataframe.groupby('categoria')['faturamento'].sum().reset_index()

# Criar gráfico de faturamento por categoria
plt.bar(df_categ_fat['categoria'], df_categ_fat['faturamento'])
plt.xlabel('Categorias')
plt.ylabel('Faturamento')
plt.title('Faturamento por Categoria')
plt.show()
```

```python
# Crie uma tabela chamada resumo_vendas com:
# - produto
# - categoria
# - quantidade_total
# - faturamento_total
df_resumo_vendas = dataframe.groupby(['produto', 'categoria']).agg(
    quantidade_total = ('quantidade', 'sum'),
    faturamento_total = ('faturamento', 'sum')
    ).reset_index()
```

```python
# Exportar a tabela `resumo_vendas` em arquivo CSV
df_resumo_vendas.to_csv(r'..\dados\resumo\resumo_vendas.csv', index=False)
```

```python
# Exibir resumo final gerado
display(df_resumo_vendas.head())
```

```python
# Qual categoria vende mais?
df_resumo_vendas = df_resumo_vendas.sort_values(by=['quantidade_total'], ascending=False)

fig_categoria, graf_categoria = plt.subplots(figsize=(8, 5))

graf_categoria.barh(df_resumo_vendas['categoria'], df_resumo_vendas['quantidade_total'])

graf_categoria.set_xlabel('Qtd. Vendida')
graf_categoria.set_ylabel('Categoria')
graf_categoria.set_title('Qtd. Vendida por Categoria')
plt.show()
```

```python
# Qual produto vende mais?
df_produtos = dataframe.groupby('produto')['quantidade'].sum().reset_index()

df_produtos = df_produtos.sort_values(by='quantidade', ascending=False)

fig_produto, graf_produto = plt.subplots(figsize=(14, 4))

graf_produto.bar(df_produtos['produto'], df_produtos['quantidade'])
graf_produto.set_xlabel('Produto')
graf_produto.set_ylabel('Qtd. Vendida')
graf_produto.set_title('Qtd. Vendida por Produto')
plt.show()
```

```python
# Qual vendedor vende mais?
df_vendedor = dataframe.groupby('vendedor')['faturamento'].sum().reset_index()

df_vendedor = df_vendedor.sort_values(by='faturamento', ascending=False)

fig_vendedor, graf_vendedor = plt.subplots(figsize=(8, 5))

graf_vendedor.bar(df_vendedor['vendedor'], df_vendedor['faturamento'])
graf_vendedor.set_xlabel('Vendedor')
graf_vendedor.set_ylabel('Faturamento')
graf_vendedor.set_title('Performance por Vendedor')
plt.show()
```

```python
# Qual mês vende mais?

# Converter a coluna 'data_venda' para o tipo datetime
dataframe['data_venda'] = pd.to_datetime(dataframe['data_venda'])

# Criar a coluna 'mes_ano'
dataframe['mes_ano'] = dataframe['data_venda'].dt.to_period('M')

df_mes = dataframe.groupby('mes_ano')['faturamento'].sum().reset_index()

df_mes = df_mes.sort_values(by='faturamento', ascending=False)

display(df_mes)
```

```python
# Existe cliente que compra mais de uma vez?
df_cliente = dataframe.groupby('cliente').agg(
    quantidade_compras = ('id_venda', 'count')
    ).reset_index()

df_cliente = df_cliente.sort_values(by='quantidade_compras', ascending=False)

display(df_cliente)
```
